```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 

df = pd.read_csv('heart_failure_clinical_records_dataset.csv')
df.head(10)
```

```python
sns.kdeplot( data = df, x = 'ejection_fraction', fill = True).set(title='Gaussian KDE (Smooth)')
```

```python
plt.figure(figsize=(6,5))

for i in range(15):
    resample = df.sample(frac=1.0, replace=True)
    sns.kdeplot( data = resample, x = 'ejection_fraction', fill = True).set(title='Gaussian KDE (Smooth)')

plt.show()
```

The values of the original ejection fraction plot that are reliable is the range of (0, 20) as there's little to no variation here, which is surprising for an endpoint. Besides one of the samples, the peak at 60 is also very reliable as well. There's variation in the 20s, the peak at 40, and the endpoint at >=80.

```python
from scipy.stats import norm

# Data and parameters
data = np.array([5, 7, 7, 8, 10])
points = [6.0, 6.5, 7.0, 8.0, 8.5]

# Gaussian KDE calculation
for x in points:
    u = (x - data) / 1.5
    fhat = np.sum(norm.pdf(u)) / 7.5
    print(fhat)
```
